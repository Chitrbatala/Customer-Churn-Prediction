```python
tenure_bins = [0, 12, 24, 48, 72]

tenure_labels = [
    "New (0-12)",
    "Early (13-24)",
    "Established (25-48)",
    "Long-term (49-72)"
]

df["tenure_group"] = pd.cut(
    df["tenure"],
    bins=tenure_bins,
    labels=tenure_labels,
    include_lowest=True
)
```

```python
categorical_cols = [
    "gender",
    "SeniorCitizen",
    "Partner",
    "Dependents",
    "PhoneService",
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies",
    "Contract",
    "PaperlessBilling",
    "PaymentMethod"
]

categorical_churn = {}

for col in categorical_cols:
    categorical_churn[col] = (
        df.groupby(col)["Churn"]
        .apply(lambda x: (x == "Yes").mean() * 100)
        .round(2)
    )

categorical_churn
```

```python
categorical_summary = []

for col in categorical_cols:
    churn_rates = (
        df.groupby(col)["Churn"]
        .apply(lambda x: (x == "Yes").mean() * 100)
    )
    
    categorical_summary.append({
        "feature": col,
        "lowest_churn": churn_rates.min(),
        "highest_churn": churn_rates.max(),
        "difference": churn_rates.max() - churn_rates.min()
    })

categorical_summary = pd.DataFrame(categorical_summary)

categorical_summary.sort_values(
    "difference",
    ascending=False
)
```
